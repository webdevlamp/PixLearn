### 1. Introduction to OpenCV
OpenCV (Open Source Computer Vision Library) is used for real-time computer vision applications.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print(f"OpenCV version: {cv2.__version__}")
```

### 2. Creating and Displaying Images

```python
black_image = np.zeros((300, 300, 3), dtype=np.uint8)
white_image = np.ones((300, 300, 3), dtype=np.uint8) * 255

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(cv2.cvtColor(black_image, cv2.COLOR_BGR2RGB))
axes[0].set_title('Black Image')
axes[1].imshow(cv2.cvtColor(white_image, cv2.COLOR_BGR2RGB))
axes[1].set_title('White Image')
plt.show()
```

### 3. Drawing Shapes

```python
canvas = np.ones((400, 400, 3), dtype=np.uint8) * 255

cv2.rectangle(canvas, (50, 50), (150, 150), (0, 0, 255), 3)
cv2.circle(canvas, (250, 100), 50, (0, 255, 0), -1)
cv2.line(canvas, (50, 250), (200, 350), (255, 0, 0), 3)
cv2.putText(canvas, 'OpenCV', (150, 300), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 0), 2)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
plt.title('Drawing Shapes')
plt.axis('off')
plt.show()
```

### 4. Color Spaces and Conversions

```python
gradient = np.zeros((200, 600, 3), dtype=np.uint8)
gradient[:, :200, 2] = np.linspace(0, 255, 200)
gradient[:, 200:400, 1] = np.linspace(0, 255, 200)
gradient[:, 400:, 0] = np.linspace(0, 255, 200)

gray = cv2.cvtColor(gradient, cv2.COLOR_BGR2GRAY)
hsv = cv2.cvtColor(gradient, cv2.COLOR_BGR2HSV)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(cv2.cvtColor(gradient, cv2.COLOR_BGR2RGB))
axes[0].set_title('BGR')
axes[1].imshow(gray, cmap='gray')
axes[1].set_title('Grayscale')
axes[2].imshow(cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB))
axes[2].set_title('HSV')
for ax in axes:
    ax.axis('off')
plt.show()
```

### 5. Image Filtering

```python
img = np.zeros((300, 300), dtype=np.uint8)
cv2.circle(img, (150, 150), 80, 255, -1)
cv2.rectangle(img, (100, 100), (200, 200), 128, -1)

blurred = cv2.GaussianBlur(img, (15, 15), 0)
edges = cv2.Canny(img, 100, 200)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(img, cmap='gray')
axes[0].set_title('Original')
axes[1].imshow(blurred, cmap='gray')
axes[1].set_title('Gaussian Blur')
axes[2].imshow(edges, cmap='gray')
axes[2].set_title('Canny Edges')
for ax in axes:
    ax.axis('off')
plt.show()
```

### 6. Thresholding

```python
gradient_img = np.tile(np.linspace(0, 255, 300, dtype=np.uint8), (300, 1))

_, binary = cv2.threshold(gradient_img, 127, 255, cv2.THRESH_BINARY)
_, binary_inv = cv2.threshold(gradient_img, 127, 255, cv2.THRESH_BINARY_INV)
adaptive = cv2.adaptiveThreshold(gradient_img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

fig, axes = plt.subplots(2, 2, figsize=(10, 10))
axes[0, 0].imshow(gradient_img, cmap='gray')
axes[0, 0].set_title('Original Gradient')
axes[0, 1].imshow(binary, cmap='gray')
axes[0, 1].set_title('Binary Threshold')
axes[1, 0].imshow(binary_inv, cmap='gray')
axes[1, 0].set_title('Binary Inverse')
axes[1, 1].imshow(adaptive, cmap='gray')
axes[1, 1].set_title('Adaptive Threshold')
for ax in axes:
    for a in ax:
        a.axis('off')
plt.show()
```

### 7. Contour Detection

```python
shapes = np.ones((300, 400, 3), dtype=np.uint8) * 255
cv2.rectangle(shapes, (50, 50), (150, 150), (0, 0, 0), -1)
cv2.circle(shapes, (280, 100), 50, (0, 0, 0), -1)
cv2.drawContours(shapes, [np.array([[200, 200], [350, 250], [250, 280]])], 0, (0, 0, 0), -1)

gray_shapes = cv2.cvtColor(shapes, cv2.COLOR_BGR2GRAY)
_, thresh = cv2.threshold(gray_shapes, 127, 255, cv2.THRESH_BINARY_INV)
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

output = shapes.copy()
for i, contour in enumerate(contours):
    area = cv2.contourArea(contour)
    perimeter = cv2.arcLength(contour, True)
    cv2.drawContours(output, [contour], 0, (0, 255, 0), 2)
    print(f"Shape {i+1}: Area={area:.0f}, Perimeter={perimeter:.0f}")

plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title('Detected Contours')
plt.axis('off')
plt.show()
```

### 8. Working with Real Images (MNIST)

```python
from tensorflow.keras.datasets import mnist

(x_train, _), (x_test, _) = mnist.load_data()

sample = x_test[0]

_, binary = cv2.threshold(sample, 127, 255, cv2.THRESH_BINARY)
contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

sample_rgb = cv2.cvtColor(sample, cv2.COLOR_GRAY2BGR)
cv2.drawContours(sample_rgb, contours, -1, (0, 255, 0), 1)

fig, axes = plt.subplots(1, 3, figsize=(10, 4))
axes[0].imshow(sample, cmap='gray')
axes[0].set_title('Original')
axes[1].imshow(binary, cmap='gray')
axes[1].set_title('Thresholded')
axes[2].imshow(cv2.cvtColor(sample_rgb, cv2.COLOR_BGR2RGB))
axes[2].set_title('With Contours')
for ax in axes:
    ax.axis('off')
plt.show()
```
